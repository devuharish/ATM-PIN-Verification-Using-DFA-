```python

def atm_pin_dfa(pin):

    state = "q0"

    for digit in pin:

        if digit not in "0123456789":
            state = "qd"
            break

        if state == "q0":
            state = "q1"

        elif state == "q1":
            state = "q2"

        elif state == "q2":
            state = "q3"

        elif state == "q3":
            state = "q4"

        elif state == "q4":
            state = "qd"

    if state == "q4":
        return "Valid PIN Format"

    else:
        return "Invalid PIN Format"
```

```python

test_pins = ["1234", "5678", "123", "12345", "12AB"]

for pin in test_pins:
    print(pin, ":", atm_pin_dfa(pin))
```

```python

import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# Create the diagram
fig, ax = plt.subplots(figsize=(12, 4))

states = {
    "q0": (1, 2),
    "q1": (3, 2),
    "q2": (5, 2),
    "q3": (7, 2),
    "q4": (9, 2),
    "qd": (5, 0)
}

# Draw states
for state, (x, y) in states.items():

    circle = Circle((x, y), 0.35,
                    fill=False, linewidth=2)
    ax.add_patch(circle)

    if state == "q4":
        inner = Circle((x, y), 0.27,
                       fill=False, linewidth=1.5)
        ax.add_patch(inner)

    ax.text(x, y, state,
            ha="center", va="center",
            fontsize=12)

# Draw transitions
for i in range(4):
    x1 = 1 + (i * 2)
    x2 = x1 + 2

    ax.annotate(
        "",
        xy=(x2 - 0.35, 2),
        xytext=(x1 + 0.35, 2),
        arrowprops=dict(arrowstyle="->", linewidth=1.5)
    )

    ax.text((x1+x2)/2, 2.25, "0-9",
            ha="center", fontsize=10)

# Start arrow
ax.annotate("", xy=(0.65, 2),
            xytext=(0.1, 2),
            arrowprops=dict(arrowstyle="->"))

ax.text(0.1, 2.3, "Start", fontsize=10)

# Extra digit from q4 to dead state
ax.annotate(
    "",
    xy=(5.35, 0.35),
    xytext=(9, 1.65),
    arrowprops=dict(arrowstyle="->",
                    connectionstyle="arc3,rad=-0.2")
)

ax.text(7.4, 0.8, "Extra digit",
        fontsize=9)

# Dead state loop
ax.annotate(
    "",
    xy=(4.75, 0.1),
    xytext=(5.25, 0.1),
    arrowprops=dict(arrowstyle="->",
                    connectionstyle="arc3,rad=2")
)

ax.text(5, -0.65, "qd = Dead State",
        ha="center", fontsize=10)

ax.set_xlim(-0.2, 10)
ax.set_ylim(-1, 3)
ax.axis("off")

plt.title("ATM PIN Verification System - DFA",
          fontsize=15)
plt.savefig("ATM_PIN_DFA_Diagram.png", dpi=300, bbox_inches="tight")

plt.show()
```
